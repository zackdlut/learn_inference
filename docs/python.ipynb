```python
# 为什么 交叉熵损失用log(x)
# y = -log(x)
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0.01, 10, 1000)
y = -np.log(x)
z = np.log(x)

plt.plot(x, y, color='blue')
plt.plot(x, z, color='red')

plt.axvline(x=0, color='green', linestyle='--')
plt.axhline(y=0, color='purple', linestyle='--')

# 标记关键点
plt.scatter([1, np.e], [0, -1], color='red')
plt.text(1, 0.1, '(1, 0)')
plt.text(np.e, -1.2, r'$(e, -1)$')

# 网格与标签
plt.grid(True, linestyle='--', alpha=0.6)
plt.xlabel('x')
plt.ylabel('y')
plt.title(r'Graph of $y=-\log(x)$')
plt.legend()

plt.show()
```
